### A2. NumPy: tính toán trên cả mảng cùng lúc
Mảng NumPy cho phép tính trên toàn bộ dữ liệu mà không cần vòng lặp (vector hóa) – nhanh và gọn.

```python
gia = np.array([250, 189, 1200, 99, 560])      # nghìn đồng
sl  = np.array([10, 25, 3, 40, 8])
doanh_thu = gia * sl
print("Doanh thu từng SP:", doanh_thu)
print("Tổng:", doanh_thu.sum(), "| TB:", doanh_thu.mean(), "| Độ lệch chuẩn:", doanh_thu.std().round(1))
print("SP có doanh thu > 3000:", np.where(doanh_thu > 3000)[0])
print("Tỷ trọng (%):", (doanh_thu / doanh_thu.sum() * 100).round(1))
```

### Bài B1. Tổng quan dữ liệu

Cho biết: số hóa đơn, khoảng thời gian dữ liệu (ngày đầu – ngày cuối), tổng doanh thu (tổng `tong_tien`, quy ra tỷ đồng),
tổng lợi nhuận gộp, giá trị đơn trung bình và điểm đánh giá trung bình.

```python
#Đếm số hoá đơn
so_hoa_don = len(df)
#Khoảng thời gian
ngay_dau = df["ngay"].min()
ngay_cuoi = df["ngay"].max()
#Tổng doanh thu (quy ra tỷ đồng)
tong_doanh_thu = df["tong_tien"].sum() / 1_000_000 #vì bảng số liệu đơn vị nghìn đồng
#Tổng lợi nhuận gộp
tong_loi_nhuan = df["loi_nhuan_gop"].sum() *1_000
#Giá trị đơn trung bình
gia_tri_don_trung_binh = df["tong_tien"].mean() *1_000
#Điểm đánh giá trung bình
diem_trung_binh = df["danh_gia"].mean()
#In kết quả
print(f"Số hoá đơn: {so_hoa_don:,}")
print(f"Khoảng thời gian: Từ {ngay_dau} đến {ngay_cuoi}")
print(f"Tổng doanh thu: {tong_doanh_thu:,.2f} tỷ đồng")
print(f"Tổng lợi nhuận gộp: {tong_loi_nhuan:,.0f} đồng")
print(f"Giá trị đơn trung bình: {gia_tri_don_trung_binh:,.0f} đồng")
print(f"Điểm đánh giá trung bình: {diem_trung_binh:,.1f}")
```

### Bài B2. Lọc dữ liệu theo yêu cầu quản lý

(a) Các hóa đơn của **khách thành viên** ở **Hà Nội** thanh toán bằng **ví điện tử** – có bao nhiêu hóa đơn, tổng tiền bao nhiêu?
(b) 5 hóa đơn có điểm đánh giá thấp nhất (hiển thị mã hóa đơn, chi nhánh, nhóm hàng, điểm đánh giá).
(c) Số hóa đơn mua từ 8 sản phẩm trở lên.

```python
#a, các hoá đơn thanh toán bằng ví điện tử của khách thành viên ở Hà Nội
dieu_kien_a = (df["loai_khach"] == "Thành viên") & (df["thanh_pho"] == "Hà Nội") & (df["thanh_toan"] == "Ví điện tử")
df_a = df[dieu_kien_a] #dùng dataframe để lấy ra các dòng thoả mãn điều kiện
so_hoa_don = len(df_a)
tong_tien = df_a["tong_tien"].sum() * 1_000 #đơn vị đang là nghìn đồng
print(f"(a) Số hoá đơn thoả mãn điều kiện câu a: {so_hoa_don}")
print(f"Tổng tiền: {tong_tien:,.0f} đồng")
```

### Bài B3. Doanh thu theo nhóm hàng

Tính tổng doanh thu, số hóa đơn và tỷ trọng doanh thu (%) của từng nhóm hàng, sắp xếp giảm dần.
Nhóm nào đóng góp nhiều nhất?

<details><summary>💡 Gợi ý</summary>

`df.groupby('nhom_hang')['tong_tien'].sum()` cho tổng; chia cho tổng chung để ra tỷ trọng.

</details>

```python
#Gom nhóm và tính tổng doanh thu, số hoá đơn
df_nhom_hang = df.groupby("nhom_hang").agg (tong_doanh_thu = ("tong_tien", "sum"), so_hoa_don = ("ma_hoa_don", "nunique"))
# nunique đếm mã hoá đơn không trùng lặp
#Tính tỷ trọng doanh thu (%)
tong_doanh_thu_tat_ca = df_nhom_hang["tong_doanh_thu"].sum()
df_nhom_hang["ty_trong_%"] = (df_nhom_hang["tong_doanh_thu"] / tong_doanh_thu_tat_ca * 100).round(2)
#Sắp xếp giảm dần theo doanh thu
df_nhom_hang = df_nhom_hang.sort_values("tong_doanh_thu", ascending=False)
#In kết quả
print ("Bảng thống kê doanh thu theo nhóm hàng:")
(df_nhom_hang)
```

### Bài B4. Cột tính toán mới

Tạo cột `muc_danh_gia` nhận giá trị "Tốt" nếu `danh_gia` ≥ 8, "Trung bình" nếu từ 6 đến dưới 8, "Kém" nếu < 6.
Đếm số hóa đơn theo từng mức và tính tỷ lệ phần trăm.

<details><summary>💡 Gợi ý</summary>

Dùng `np.where(dieu_kien, gia_tri_dung, gia_tri_sai)` lồng nhau, hoặc `pd.cut`.

</details>

```python
# Tạo cột mức đánh giá
df["muc_danh_gia"] = np.where (df["danh_gia"] >= 8, "Tốt", np.where(df["danh_gia"] >= 6, "Trung bình", "Kém"))
#Đếm hoá đơn theo từng mức đánh giá
df_danh_gia = df.groupby("muc_danh_gia").agg(so_hoa_don=("ma_hoa_don", "nunique"))
#Tính tỷ lệ phần trăm
df_danh_gia["ty_le_%"] = (df_danh_gia["so_hoa_don"] / df_danh_gia["so_hoa_don"].sum() *100)
print ("Phân loại đơn hàng theo đánh giá:")
(df_danh_gia)
```

### Bài B5. Xuất báo cáo chi nhánh

Với mỗi thành phố, lọc dữ liệu và ghi ra tệp CSV riêng `bao_cao_<thành phố>.csv` (dùng vòng lặp).
In số dòng của mỗi tệp đã ghi.

```python
#Lấy danh sách thành phố
danh_sach_thanh_pho = df["thanh_pho"].unique()
for tp in danh_sach_thanh_pho:
    #lọc dữ liệu theo thành phố
    df_tp = df[df["thanh_pho"] == tp]
    #tạo tên file
    ten_file = f"bao_cao_{tp}.csv"
    #Ghi dữ liệu
    df_tp.to_csv(ten_file, index=False, encoding="utf-8-sig")
    print(f"{ten_file}: {len(df_tp)} dòng")
```
