### A2. Tính giá bán sau chiết khấu và thuế VAT
**Tình huống:** Cửa hàng bán một chiếc tai nghe giá niêm yết 850.000 đ, giảm 15%, sau đó cộng VAT 8%. Khách mua 3 chiếc.

**Bước 1 – Khai báo biến.** Tên biến viết thường, nối bằng dấu `_`, đặt tên có nghĩa.

```python
gia_niem_yet = 850_000      # đồng (dấu _ giúp dễ đọc số lớn)
ty_le_giam = 0.15           # 15%
vat = 0.08                  # 8%
so_luong = 3
print(type(gia_niem_yet), type(ty_le_giam), type(so_luong))
```

**Bước 2 – Tính toán** với các toán tử `+ - * /`.

```python
gia_sau_giam = gia_niem_yet * (1 - ty_le_giam)
gia_sau_vat = gia_sau_giam * (1 + vat)
thanh_tien = gia_sau_vat * so_luong
print(gia_sau_giam, gia_sau_vat, thanh_tien)
```

**Bước 3 – Định dạng kết quả.** f-string `f"...{bien:,.0f}..."` chèn giá trị vào chuỗi;
`:,.0f` = dấu phẩy ngăn cách hàng nghìn, 0 chữ số thập phân. Thay `,` bằng `.` để đúng kiểu Việt Nam.

```python
def vnd(x):
    return f"{x:,.0f}".replace(",", ".") + " đ"

print("Giá sau giảm :", vnd(gia_sau_giam))
print("Giá sau VAT  :", vnd(gia_sau_vat))
print("Thành tiền   :", vnd(thanh_tien))
```

### A3. Phép chia lấy nguyên `//`, chia lấy dư `%` và lũy thừa `**`
**Tình huống:** Kho có 1.250 chai nước, mỗi thùng đóng 24 chai. Được bao nhiêu thùng, dư bao nhiêu chai?

```python
tong_chai = 1250
chai_moi_thung = 24
so_thung = tong_chai // chai_moi_thung
du = tong_chai % chai_moi_thung
print(f"Đóng được {so_thung} thùng, còn dư {du} chai")
```

**Lãi kép:** Gửi 100 triệu đồng, lãi suất 6%/năm, sau 5 năm được bao nhiêu? Công thức: `A = P × (1 + r)^n`.

```python
P, r, n = 100_000_000, 0.06, 5
A = P * (1 + r) ** n
print("Số tiền sau", n, "năm:", vnd(A))
print("Tiền lãi:", vnd(A - P))
```

### A4. Xử lý chuỗi: chuẩn hóa dữ liệu khách hàng
Dữ liệu nhập tay thường có khoảng trắng thừa, viết hoa lung tung. Các phương thức chuỗi hay dùng:
`strip()`, `lower()`, `upper()`, `title()`, `replace()`, `split()`, `len()`.

```python
ten_kh = "   nguyễN   văn  AN  "
email = " An.Nguyen@Gmail.COM "
ten_chuan = " ".join(ten_kh.split()).title()
email_chuan = email.strip().lower()
print(repr(ten_chuan), repr(email_chuan))
print("Tên miền email:", email_chuan.split("@")[1])
print("Số ký tự của tên:", len(ten_chuan))
```

Cắt chuỗi (slicing) `chuoi[bat_dau:ket_thuc]` – vị trí bắt đầu từ 0:

```python
ma_don = "HCM-2025-000123"
print("Chi nhánh:", ma_don[:3])
print("Năm:", ma_don[4:8])
print("Số thứ tự:", int(ma_don[-6:]))
```
