## Phần A. Bài tập có hướng dẫn

### A1. Phân hạng khách hàng bằng if/elif/else
**Quy tắc:** chi tiêu năm ≥ 50 triệu → *Kim cương*; ≥ 20 triệu → *Vàng*; ≥ 5 triệu → *Bạc*; còn lại → *Thường*.
Python xét điều kiện từ trên xuống, gặp điều kiện đúng đầu tiên thì dừng.

```python
chi_tieu = 23_500_000
if chi_tieu >= 50_000_000:
    hang = "Kim cương"
elif chi_tieu >= 20_000_000:
    hang = "Vàng"
elif chi_tieu >= 5_000_000:
    hang = "Bạc"
else:
    hang = "Thường"
print("Hạng khách hàng:", hang)
```

Kết hợp điều kiện bằng `and`, `or`, `not`:

```python
gia_tri_don, la_thanh_vien = 450_000, True
mien_phi_ship = gia_tri_don >= 500_000 or (la_thanh_vien and gia_tri_don >= 300_000)
print("Miễn phí vận chuyển?", mien_phi_ship)
```

### A2. List và vòng lặp for: tổng hợp doanh số tuần
`list` lưu một dãy giá trị có thứ tự. `for` duyệt lần lượt từng phần tử.

```python
doanh_so = [12.5, 15.2, 9.8, 18.4, 21.0, 30.5, 27.3]   # triệu đồng, Thứ 2 → Chủ nhật
ngay = ["T2", "T3", "T4", "T5", "T6", "T7", "CN"]

tong = 0
for ds in doanh_so:
    tong += ds
print(f"Tổng: {tong:.1f} – Trung bình: {tong / len(doanh_so):.2f} triệu/ngày")
print("Cao nhất:", max(doanh_so), "– Thấp nhất:", min(doanh_so), "– sum():", sum(doanh_so))

for n, ds in zip(ngay, doanh_so):          # zip ghép 2 list
    nhan = "⬆ vượt 20tr" if ds > 20 else ""
    print(f"{n}: {ds:5.1f} {nhan}")
```

`enumerate()` cho cả chỉ số; `range()` tạo dãy số; list comprehension viết vòng lặp gọn trong một dòng:

```python
for i, ds in enumerate(doanh_so, start=1):
    if i > 3: break
    print(i, ds)

ngay_vuot = [n for n, ds in zip(ngay, doanh_so) if ds > 20]
doanh_so_usd = [round(ds * 1e6 / 25_350) for ds in doanh_so]
print(ngay_vuot, doanh_so_usd)
```

### A3. Dictionary: danh mục sản phẩm
`dict` lưu cặp **khóa → giá trị**, tra cứu nhanh theo khóa. Một list các dict là cách biểu diễn bảng dữ liệu đơn giản.

```python
gia = {"SP01": 250_000, "SP02": 189_000, "SP03": 1_200_000}
gia["SP04"] = 99_000                 # thêm
gia["SP02"] = 175_000                # sửa
print(gia["SP03"], gia.get("SP99", "không có"))
for ma, g in gia.items():
    print(ma, f"{g:,}")

don_hang = [
    {"ma": "DH1", "kh": "An",   "sp": "SP01", "sl": 2},
    {"ma": "DH2", "kh": "Bình", "sp": "SP03", "sl": 1},
    {"ma": "DH3", "kh": "An",   "sp": "SP04", "sl": 5},
]
for d in don_hang:
    d["thanh_tien"] = gia[d["sp"]] * d["sl"]
print(don_hang)
```

Đếm/tổng theo nhóm bằng dict – ý tưởng đứng sau `groupby` của Pandas ở buổi 5:

```python
tong_theo_kh = {}
for d in don_hang:
    tong_theo_kh[d["kh"]] = tong_theo_kh.get(d["kh"], 0) + d["thanh_tien"]
print(tong_theo_kh)
print("Khách hàng khác nhau (set):", {d["kh"] for d in don_hang})
```

### A4. Hàm: đóng gói công thức KPI
Hàm giúp viết một lần, dùng nhiều lần. Có thể có **tham số mặc định** và **trả về nhiều giá trị** (tuple).

```python
def kpi(doanh_thu, gia_von, chi_phi_hd=0):
    """Trả về (lợi nhuận gộp, lợi nhuận ròng, biên lợi nhuận ròng)."""
    ln_gop = doanh_thu - gia_von
    ln_rong = ln_gop - chi_phi_hd
    bien = ln_rong / doanh_thu if doanh_thu else 0
    return ln_gop, ln_rong, bien

gop, rong, bien = kpi(1200, 780, 250)
print(gop, rong, f"{bien:.1%}")
print(kpi(doanh_thu=500, gia_von=300))   # dùng giá trị mặc định
```

Hàm ngắn có thể viết bằng `lambda`, thường dùng để làm khóa sắp xếp:

```python
sp = [("Áo", 250_000, 120), ("Quần", 420_000, 45), ("Mũ", 90_000, 300)]
print(sorted(sp, key=lambda x: x[1] * x[2], reverse=True))   # xếp theo doanh thu
```

### A5. while và try/except: nhập liệu an toàn
`while` lặp đến khi điều kiện sai. `try/except` bắt lỗi để chương trình không dừng đột ngột.

```python
du_lieu_nhap = ["abc", "-5", "12"]       # giả lập người dùng nhập 3 lần
i = 0
while True:
    gia_tri = du_lieu_nhap[i]; i += 1
    try:
        sl = int(gia_tri)
        if sl <= 0:
            raise ValueError("số lượng phải dương")
        break
    except ValueError as e:
        print(f"'{gia_tri}' không hợp lệ ({e}). Nhập lại.")
print("Số lượng hợp lệ:", sl)
```

## Phần B. Bài tập tự làm

### Bài B1. Hoa hồng nhân viên bán hàng

Hoa hồng tính theo doanh số tháng: dưới 100 triệu → 2%; 100–300 triệu → 3,5%; trên 300 triệu → 5% và thưởng thêm 2 triệu.
Viết hàm `hoa_hong(doanh_so)` trả về số tiền hoa hồng, rồi tính cho danh sách
`nhan_vien = {"Lan": 85e6, "Minh": 210e6, "Tuấn": 345e6, "Hà": 300e6}`.

```python
#hàm hoa hồng theo doanh số
def hoa_hong(doanh_so):
    if doanh_so < 100_000_000:
        hoa_hong = doanh_so * 0.02
    elif doanh_so <= 300_000_000:
        hoa_hong = doanh_so * 0.035
    elif doanh_so > 300_000_000:
        hoa_hong = doanh_so * 0.05 + 2_000_000
    return hoa_hong
#doanh số nhân viên (e6 là 10 mũ 6)
nhan_vien = {
    "Lan": 85e6,
    "Minh": 210e6,
    "Tuấn": 345e6,
    "Hà": 300e6}
#tính danh sách nhân viên
print ("Bảng hoa hồng nhân viên")
for ten, doanh_so in nhan_vien.items():
    tien_nhan_duoc = hoa_hong(doanh_so)
    print (f"{ten}: {tien_nhan_duoc:,.0f} đồng")
```

### Bài B2. Phân tích danh sách đơn hàng

Cho `gia_tri_don = [320, 1250, 85, 640, 2100, 450, 99, 780, 1500, 230]` (nghìn đồng). Hãy tính:
số đơn, tổng, trung bình, số đơn ≥ 500 và tỷ lệ của chúng (%), và một list mới chỉ gồm các đơn < 100 (đơn nhỏ).

```python
# Giá trị đơn hàng: đơn vị: nghìn đồng
gia_tri_don = [320, 1250, 85, 640, 2100, 450, 99, 780, 1500, 230]

# Số lượng đơn hàng (hàm len để đếm số phần tử)
so_don = len(gia_tri_don)

# Tổng giá trị các đơn
tong_gia_tri = sum(gia_tri_don)

# Giá trị trung bình
trung_binh = sum(gia_tri_don) / so_don

# Số dơn lớn hơn bằng 500 và tính tỷ lệ
cac_don_lon = [don for don in gia_tri_don if don >= 500]
so_don_lon = len(cac_don_lon)
ty_le_don_lon = (so_don_lon / so_don) * 100

#List mới các đơn nhỏ hơn 500
cac_don_nho = [don for don in gia_tri_don if don < 100]

#In kết quả
print (f"Số lượng đơn hàng: {so_don} đơn")
print (f"Tổng giá trị: {tong_gia_tri:,.0f} nghìn đồng")
print (f"Giá trị trung bình: {trung_binh:,.1f} nghìn đồng/đơn")
print (f"Số đơn lớn hơn hoặc bằng 500: {cac_don_lon} - Tỷ lệ: {ty_le_don_lon:,.1f}")
print (f"Danh sách các đơn nhỏ hơn 100: {cac_don_nho}")
```

### Bài B3. Giỏ hàng

Cho bảng giá `gia = {"Cà phê": 29000, "Trà sữa": 35000, "Bánh mì": 20000, "Nước suối": 10000}` và giỏ hàng
`gio = [("Cà phê", 2), ("Bánh mì", 3), ("Trà đào", 1), ("Nước suối", 4)]`.
Tính tổng tiền; món không có trong bảng giá thì in cảnh báo và bỏ qua. Nếu tổng ≥ 150.000 đ giảm 10%.

```python
#Khai báo giá
gia = {"Cà phê": 29_000, "Trà sữa": 35_000, "Bánh mì": 20_000, 'Nước suối': 10_000}
#Khai báo giỏ
gio = [("Cà phê", 2), ("Bánh mì", 3), ("Trà đào", 1), ("Nước suối", 4)]
tong_tien = 0
#Duyệt giỏ hàng
for mon, so_luong in gio:
    if mon in gia:
        #NẾU CÓ
        thanh_tien = gia[mon] * so_luong
        tong_tien += thanh_tien # tương đương tổng tiền = tổng tiền + thành tiền
        #NẾU KHÔNG CÓ
    else:
        print (f"CẢNH BÁO: {mon} không có trong giỏ hàng, bỏ qua")
print (f"Tổng tiền trước giảm giá: {tong_tien:,} đồng")
if tong_tien >= 150_000:
    giam_gia = tong_tien * 0.10
else:
    giam_gia = 0
thanh_toan = tong_tien - giam_gia
print (f"Giảm giá: {giam_gia:,.0f} đồng")
print (f"Tổng tiền phải trả: {thanh_toan:,.0f} đồng")
```

### Bài B4. Thống kê theo nhóm

Cho danh sách giao dịch `gd = [("Điện tử", 12.5), ("Thời trang", 3.2), ("Điện tử", 8.1), ("Thực phẩm", 1.5),
("Thời trang", 4.8), ("Điện tử", 15.0), ("Thực phẩm", 2.2)]` (triệu đồng).
Dùng dict để tính tổng doanh thu và số giao dịch mỗi nhóm hàng; in nhóm có doanh thu cao nhất.

```python
#Khai báo giao dịch. Đơn vị: triệu đồng
gd = [("Điện tử", 12.5), ("Thời trang", 3.2), ("Điện tử", 8.1), ("Thực phẩm", 1.5), ("Thời trang", 4.8), ("Điện tử", 15.0), ("Thực phẩm", 2.2)]
#Tạo từ điển rỗng
doanh_thu = {}
so_giao_dich = {}
#Duyệt giao dịch
for nhom, tien in gd:
    #Nếu không xuất hiện
    if nhom not in doanh_thu:
        doanh_thu[nhom] = 0
        so_giao_dich[nhom] = 0
    doanh_thu[nhom] += tien
    so_giao_dich[nhom] += 1
#In kết quả giao dịch theo nhóm
for nhom in doanh_thu:
    print (f"{nhom}: Doanh thu = {doanh_thu[nhom]} triệu đồng")
    print (f"số giao dịch = {so_giao_dich[nhom]}")
#Tìm nhóm có doanh thu cao nhất
nhom_cao_nhat = max(doanh_thu, key = doanh_thu.get)
print(f"Nhóm có doanh thu cao nhất: {nhom_cao_nhat}")
print (f"Tổng doanh thu: {doanh_thu[nhom_cao_nhat]} triệu đồng")
```

### Bài B5. Kế hoạch tiết kiệm với while

Một startup có 500 triệu đồng tiền mặt, mỗi tháng tiêu (burn) 65 triệu đồng và chi phí tăng 3% mỗi tháng.
Dùng `while` để tìm số tháng công ty còn hoạt động được trước khi hết tiền (runway).

```python
tien_mat = 500
burn = 65
thang = 0
while tien_mat > burn:
    tien_mat -= burn
    thang += 1
    burn *= 1.03
print("Số tháng công ty còn hoạt động", thang, "tháng")
print(f"Số tiền còn lại: {tien_mat:,.2f} triệu đồng")
```
